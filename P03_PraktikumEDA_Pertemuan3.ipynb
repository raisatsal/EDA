## Bagian D — Menandai data pribadi

| Kategori | Artinya |
|---|---|
| **pengenal langsung** | Sendirian sudah menunjuk ke satu orang |
| **quasi-identifier** | Sendirian aman, gabungannya menunjuk ke satu orang |
| **data spesifik** | Bocornya merugikan lebih berat (kesehatan, keuangan pribadi, data anak) |
| **bukan data pribadi** | Tidak menunjuk ke orang; inilah bahan analisis |

```python
# TODO D1 — susun tabel penandaan Anda sendiri

PENANDAAN = {
    # Pengenal langsung
    'nama_lengkap': ('pengenal langsung', 'hash'),
    'nik_dummy': ('pengenal langsung', 'hash'),
    'no_hp': ('pengenal langsung', 'hash'),
    'email': ('pengenal langsung', 'hash'),
    'alamat_jalan': ('pengenal langsung', 'hapus'),
    'nama_usaha': ('pengenal langsung', 'hash'),

    # Quasi-identifier
    'id_peserta': ('quasi-identifier', 'pertahankan'),
    'kelurahan': ('quasi-identifier', 'generalisasi'),
    'kecamatan': ('quasi-identifier', 'generalisasi'),
    'kota': ('quasi-identifier', 'pertahankan'),
    'tanggal_lahir': ('quasi-identifier', 'generalisasi'),
    'jenis_kelamin': ('quasi-identifier', 'generalisasi'),
    'pendidikan_terakhir': ('quasi-identifier', 'pertahankan'),

    # Data spesifik
    'status_disabilitas': ('data spesifik', 'generalisasi'),
    'jumlah_anak': ('data spesifik', 'generalisasi'),
    'penghasilan_pribadi_bulanan': ('data spesifik', 'generalisasi'),
    'omzet_usaha_bulanan': ('data spesifik', 'generalisasi'),

    # Kolom lain akan dianggap bukan data pribadi + pertahankan
}

# Bangun DataFrame penandaan
baris_tanda = []

for kolom in df.columns:
    if kolom in PENANDAAN:
        kategori, tindakan = PENANDAAN[kolom]
    else:
        kategori, tindakan = 'bukan data pribadi', 'pertahankan'

    baris_tanda.append({
        'nama_kolom': kolom,
        'kategori': kategori,
        'tindakan': tindakan
    })

tanda = pd.DataFrame(baris_tanda)

# Tampilkan tabel penandaan
display(tanda)

# Cetak jumlah kolom per kategori
print("\nJumlah kolom per kategori:")
print(tanda['kategori'].value_counts())
```

## Bagian E — Menerapkan tiga tindakan

```python
# TODO E1 — HAPUS

# Cari kolom yang tindakannya adalah 'hapus'
kolom_hapus = [
    kolom for kolom, (_, tindakan) in PENANDAAN.items()
    if tindakan == 'hapus'
]

# Buat DataFrame baru tanpa kolom tersebut
aman = df.drop(columns=kolom_hapus)

# Cetak jumlah kolom yang dibuang dan yang tersisa
print("Kolom yang dibuang:", len(kolom_hapus))
print("Kolom yang tersisa:", aman.shape[1])
print("Nama kolom yang dibuang:", kolom_hapus)
```

```python
# TODO E2 — HASH

import hashlib
import numpy as np

GARAM = 'ganti-dengan-kalimat-rahasia-tim'

def samarkan(nilai, garam):
    # Jika nilai kosong, tetap kosong
    if pd.isna(nilai):
        return np.nan

    # Gabungkan garam + nilai lalu lakukan SHA-256
    teks = garam + str(nilai)
    hasil_hash = hashlib.sha256(teks.encode('utf-8')).hexdigest()

    # Ambil 12 karakter pertama
    return hasil_hash[:12]


# Cari kolom yang tindakannya adalah 'hash'
kolom_hash = [
    kolom for kolom, (_, tindakan) in PENANDAAN.items()
    if tindakan == 'hash'
]

# Terapkan hash ke kolom yang dipilih
for kolom in kolom_hash:
    aman[kolom] = aman[kolom].apply(
        lambda x: samarkan(x, GARAM)
    )

print("Kolom yang di-hash:")
print(kolom_hash)

print("\nContoh hasil:")
display(aman[kolom_hash].head())

GARAM = 'praktikum-tim-2025'
```

```python
# TODO E3 — buktikan hash bersifat konsisten dan tidak terbaca

nilai1 = "RAYYIS"
nilai2 = "RAYYIS"
nilai3 = "ANSHAR"

# a. Hash dua kali nilai yang sama
hash1 = samarkan(nilai1, GARAM)
hash2 = samarkan(nilai2, GARAM)

print("Hash nilai yang sama:")
print("Hash 1:", hash1)
print("Hash 2:", hash2)
print("Apakah identik?", hash1 == hash2)

# b. Hash dua nilai berbeda
hash3 = samarkan(nilai3, GARAM)

print("\nHash nilai yang berbeda:")
print("Hash RAYYIS :", hash1)
print("Hash ANSHAR :", hash3)
print("Apakah berbeda?", hash1 != hash3)
```

**Jawaban Anda:** kalau `GARAM` ikut ter-*commit* ke GitHub, apa yang masih terlindungi dan apa yang tidak?

```python
# TODO E4 — GENERALISASI

ACUAN = pd.Timestamp('2025-09-01')

usia = ACUAN.year - aman['tanggal_lahir'].dt.year
usia -= (
    (aman['tanggal_lahir'].dt.month > ACUAN.month) |
    ((aman['tanggal_lahir'].dt.month == ACUAN.month) &
     (aman['tanggal_lahir'].dt.day > ACUAN.day))
).astype(int)

aman['kelompok_umur'] = pd.cut(
    usia,
    bins=[-float('inf'), 24, 34, 44, 54, float('inf')],
    labels=['<25', '25-34', '35-44', '45-54', '55+']
)

aman = aman.drop(columns=['tanggal_lahir', 'kelurahan'])

aman['kecamatan'] = aman['kecamatan'].str.strip().str.title()

print(aman['kelompok_umur'].value_counts())
print("Jumlah kecamatan unik:", aman['kecamatan'].nunique())
```

## Bagian F — Memeriksa sisa risiko

Menghapus nama belum tentu cukup. Untuk setiap kombinasi quasi-identifier, hitung berapa orang yang punya kombinasi sama. Nilai terkecil disebut **k**. Target kelas ini: **k ≥ 5**.

```python
# TODO F1 — fungsi cek_k

def cek_k(data, kolom):
    grup = data.groupby(kolom, dropna=False).size()
    return pd.Series({
        'k': grup.min(),
        'grup_1_orang': (grup == 1).sum(),
        'total_grup': len(grup)
    })

cek_k(aman, ['kota', 'kelompok_umur'])
```

```python
# TODO F2 — ukur risiko SEBELUM penanganan

cek_k(
    df,
    ['kelurahan', 'jenis_kelamin', 'pendidikan_terakhir', 'tanggal_lahir']
)
```

```python
# TODO F4 — TANTANGAN: capai k >= 5

kolom_qi = ['kecamatan', 'jenis_kelamin', 'kelompok_umur']

hasil_f4 = cek_k(aman, kolom_qi)

print(hasil_f4)
```
